```python
def team_qualify(ranking):
        if ranking <= 5:
            return 'Qualified'
        else:
            return 'Not Qualified'

team_qualify(3)
```

```python
team_pt['Qualification'] = team_pt['Ranking'].apply(team_qualify)
```

```python
team_pt
```

```python
team_pt['Points'].apply(np.sum)
```

```python
def point_pct(value):
    return((value/sum(value)*100))

team_pt['point_pct'] = point_pct(team_pt['Points'])
team_pt['tot_point'] = team_pt['Points'].sum()
```
